# EduPro EDA

Load the cleaned merged dataset and inspect platform coverage, missingness, category mix, and revenue distributions.

```python
import pandas as pd
import plotly.express as px

data = pd.read_csv('../data/edupro_merged_cleaned.csv')
data.head()
```

```python
data.describe(include='all').T.head(25)
```

```python
category_summary = data.groupby('CourseCategory', as_index=False).agg(Enrollments=('TransactionID','count'), Revenue=('Amount','sum')).sort_values('Revenue', ascending=False)
px.bar(category_summary, x='Revenue', y='CourseCategory', orientation='h', title='Revenue by category')
```
